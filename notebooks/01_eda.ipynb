# Transaction Data Exploration
Inspect the example transaction data and its label balance.

```python
from pathlib import Path
import os
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "src").is_dir():
    if (PROJECT_ROOT.parent / "src").is_dir():
        PROJECT_ROOT = PROJECT_ROOT.parent
    else:
        PROJECT_ROOT = PROJECT_ROOT / "fraud-detection-mlops"
os.chdir(PROJECT_ROOT)
sys.path.insert(0, str(PROJECT_ROOT))
DATA_PATH = PROJECT_ROOT / "data/sample/sample_transactions.csv"
transactions = pd.read_csv(DATA_PATH)
transactions.head()
```

```python
transactions.info()
transactions.describe(include='all').T
```

```python
label_counts = transactions['is_fraud'].value_counts().rename(index={0: 'legitimate', 1: 'fraud'})
label_counts, (label_counts / len(transactions)).round(3)
```
