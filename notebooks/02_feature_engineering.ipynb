# Feature Engineering
Separate identifiers and labels from model inputs, then inspect the training-only preprocessing pipeline.

```python
from pathlib import Path
import os
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "src").is_dir():
    if (PROJECT_ROOT.parent / "src").is_dir():
        PROJECT_ROOT = PROJECT_ROOT.parent
    else:
        PROJECT_ROOT = PROJECT_ROOT / "fraud-detection-mlops"
os.chdir(PROJECT_ROOT)
sys.path.insert(0, str(PROJECT_ROOT))
from src.data_preprocessing import load_transactions
from src.feature_engineering import build_preprocessor, split_features_target

DATA_PATH = PROJECT_ROOT / "data/sample/sample_transactions.csv"
transactions = load_transactions(DATA_PATH)
features, target = split_features_target(transactions)
features.head()
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    features, target, test_size=0.25, random_state=42, stratify=target
)
preprocessor = build_preprocessor()
x_train_ready = preprocessor.fit_transform(x_train)
x_test_ready = preprocessor.transform(x_test)
x_train_ready.shape, x_test_ready.shape
```

The encoder and scaler are fitted only on the training split. The estimator pipeline in `src.train` keeps these transformations together with the classifier for inference.
